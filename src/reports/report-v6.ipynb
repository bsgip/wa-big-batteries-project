## Executive Summary

The WA battery fleet now meets about 15% of evening-peak demand, three times its share a year earlier. On days when AEMO identified a tight supply/demand balance, batteries were generally well prepared: they held charge ahead of the system peak and discharged to help manage it. The risks we see are emerging rather than binding: weaker charge in winter, batteries cancelling each other out just before the peak, and evening price forecasts that tend to run high.

- Reliance has tripled. Batteries met 5.1% of demand in the evening obligation window (ESROI, 17:30–21:30) in Dec 2024–Sep 2025, rising to 14.6% in Dec 2025–Sep 2026. Evening demand fell 3% over the same period, so the rise comes from more batteries being used harder.
- Charge at the peak is almost always adequate. Median fleet state of charge (SOC) at 17:30 is 71%; it falls below 50% on 11% of days and below 40% on 4%. On the 7 CC2T and 14 AEMO-advisory stress days with reliable data, SOC was within the normal range.
- But it is usually below what capacity credits assume. Delivering credited capacity for 4 hours requires about 78% fleet SOC. The median day falls short of this, and in winter 2026 (May–June) the median dipped to about 45%.
- Behaviour largely follows the forecasts. In the evening window, batteries' cheapest discharge offer sits close to the best price forecast for the rest of the window (median +\$1/MWh), which is expected behaviour under the ERA bidding guideline.
- Owners behave differently. Synergy (regulated) offers discharge at about \$115/MWh (noting this does not account for the full bid stack), runs a morning cycle and provides most FCESS. Neoen offers at about \$239/MWh and cycles once a day. Synergy's batteries set most WEM prices.
- Batteries now set WEM prices more than 90% of the time. This makes forecasts, offers and outcomes circular: battery behaviour shapes the prices that batteries respond to.
- Evening forecasts are less reliable than midday ones. Midday forecast error stays stable as lead time increases, whereas evening forecasts over-estimate prices more the further ahead they look.
- Batteries partly cancel each other out. Some batteries charge while others discharge, so 14–28% of fleet activity never reaches the grid. Cancelling out is highest just before the evening window (about a third of activity at 16:00–17:00) and overnight, and lowest during the window itself. It is more common among Synergy’s batteries, and cancelling out between owners has risen slightly over time.

**Implications for the next steps of this project:** There is no evidence yet of a systemic energy shortfall, and on the days AEMO flagged as tight, batteries held enough charge. D3 therefore focuses on “unwatched days”: days when AEMO did not expect a shortfall or might not be watching closely, such as weekends, winter, mornings, late evenings, or days when conditions change after 8am. The method first compares actual battery behaviour with a rule-based bidding benchmark, to understand where batteries depart from prescribed behaviour and whether those departures help or hurt system adequacy. The results could inform the rule changes now under review: the ERA’s bidding guideline review and the CC2T Review.

It then builds a data-trained model of actual behaviour to project the system forward under AEMO’s 2026 ESOO scenarios (fleet growth, coal retirement, longer duration requirements), with and without the proposed LOR-SOC obligation. D3 will identify when battery adequacy risk on unwatched days starts to rise and deliver three tested warning flags.

## Defining stress days
Analysing system adequacy requires a definition of a "stress day", which can be approached from several perspectives. So far, we have four definitions of past stress days:

- CC2T review: 8 listed stress event dates
- AEMO market advisories (LRC): 61 advisories on 48 dates
- Spare capacity below 750 MW in at least one interval: 35 dates
- Days in the top 10% of peak demand

*Spare capacity = IN-SERVICE capacity - energy injection dispatch*

### Stress days frequency

Both the AEMO market advisory and the CC2T review data show that summer has the most stress days. Events under both definitions mostly start around 3–4 pm, a couple of hours before the evening peak and the start of the ESROI. ### Battery charging synchronisation

We measure how much of the batteries' activity cancels out (some batteries charging while others discharge) rather than reaching the grid:

$$
\text{Cancelling out (\%)} = \left(1 - \frac{\text{Energy reaching the grid}}{\text{All battery activity}}\right) \times 100
$$

- **Energy reaching the grid (MWh):** in each 5-minute interval, the absolute value of the fleet's total discharging minus its total charging, converted to energy (MW × 5 minutes = MWh) and summed over the month (or hour of day).
- **All battery activity (MWh):** in each 5-minute interval, the fleet's total discharging plus its total charging, converted to MWh and summed in the same way.

For example, if for one hour battery A discharges 100 MW while battery B charges 30 MW, total activity is 130 MW and 70 MW reaches the grid, so 1 − 70 ÷ 130 = 46% cancels out.

# D3 $\cdot$ Storage Reserve Adequacy Risk Indicators

## Project context: a changing system 

WA’s grid-scale battery fleet has grown from about has grown about 26-fold since 2023, to about 6,000 MWh, and batteries now hold about a fifth of all capacity credits in the WEM. Batteries are credited as if they can deliver their credited MW for the full 4-hour evening window. Their obligation is checked interval by interval, and refunds apply only once a battery is empty. The system around them is changing quickly: WEM coal capacity falls from 1,173 MW in 2026-27 to zero by 2031-32, new batteries face 6- and 7-hour duration requirements from 2027-28, and AEMO forecasts winter-morning shortfalls from 2029-30. 

AEMO’s latest Quarterly Energy Dynamics (Q3 2026) shows the WEM averaging $120.72/MWh, up 19% on a year earlier and the highest quarter on record, while NEM prices fell. Battery charging and discharging both doubled, batteries supplied 93% of FCESS, and FCESS costs fell 98%. Coal output fell 22% on outages, gas rose 12% and wind fell 14%. 

The rules governing batteries are being rewritten. Energy Policy WA’s Capability Class 2 Technologies Review ran a working group from October 2025 to April 2026, consulted in August 2026, published its outcomes on 30 September, and released an exposure draft of amending rules in November 2026. Submissions close on 10 December 2026 and the battery provisions commence on 1 October 2027. The review keeps the current accreditation and evening-window methods. It adds a state-of-charge mandate: where AEMO judges total battery charge may be insufficient for the evening, it declares a low reserve condition by 8am and sets a uniform percentage of available charge that every battery must hold. AEMO enforces the mandate through dispatch constraints, blocking discharge below the mandate before the window and blocking charging during the declared period, and refunds double while it applies. Submissions and working group discussion raised several concerns: 

- whether the trigger would catch events that build without warning (none was issued on 25 August 2025); 
- whether a uniform percentage suits batteries of different durations; 
- whether mandated charging could itself raise prices or create stress; 
- whether planning models that assume perfect battery behaviour are a sound basis for the design; 
- whether the review’s stress days were selectively chosen; 
- whether the evening window should follow actual peak timing; 
- whether doubling refunds is justified. 

In parallel, the ERA is reviewing its bidding guideline, which requires battery offers to reflect forecast prices, and is asking how far ahead batteries may look and whether refund risk may enter charging offers. 

Our discussions with operators and AEMO adds practical insight. Battery operators say they follow the bidding guideline and AEMO’s forecasts closely, and cycle for both the morning and evening peaks. AEMO notes that batteries have removed minimum-demand risk for now, that how they plan and trade in aggregate will be a primary driver of reliability even at today’s fleet size, and that current statistics alone cannot predict a changing system. Further rule reform is being discussed beyond the published consultations. 

The new mandate relies on AEMO seeing a problem coming. Our project will try to examine the days when AEMO might not see a problem coming, and whether the bidding rules help or hinder on those days. 

Sources: AEMO QED Q3 2026; CC2T Review documents, including the Information Paper and Exposure Draft.

## Understanding batteries' ideal response to the price forecast

To model how batteries will behave in the future, we first examined how they currently respond to the price forecast. The Offer Construction Guideline allows ESRs to follow an optimal pricing strategy that accounts for their opportunity cost and their obligation to preserve energy for the ESROI.

### Ideal response modelling

As the Offer Construction Guideline is open to interpretation, we propose the following model of a battery's ideal response to the price forecast.

- The AEMO pre-dispatch price forecast is divided into two categories: the *Near* forecast and the *Far* forecast. The *Near* forecast covers the next hour, whereas the *Far* forecast is a set of windows: 1–2 h ahead, 1–4 h ahead and 1–6 h ahead.
- Comparing the *Near* and *Far* forecasts forms the basis for determining a battery's ideal response.
- The weighted average cost of the energy stored in the battery is another factor in determining the ideal response. We therefore propose two rules to model a battery's ideal response.

1. **Rule 1: the battery acts only on the comparison of the *Near* and *Far* forecasts**
    - Discharge: *Near* > *Far*
    - Charge: *Near* < 0.85 $\cdot$ *Far*, where 0.85 is the round-trip efficiency
    - Hold: otherwise
2. **Rule 2: the battery acts on the price forecast and the weighted average cost**
    - Discharge: *Near* > *Far* **and** *Near* > *Avg cost*
    - Charge: *Near* < 0.85 $\cdot$ *Far*, where 0.85 is the round-trip efficiency
    - Hold: otherwise

*Near*, *Far* and *Avg cost* are defined as follows.

1. **Near forecast.** For each trading interval, *Near* is the average forecast price over the next hour (the next two trading intervals). It uses the forecast published one trading interval earlier.

2. **Far forecast.** For each trading interval, *Far* is calculated for three forecast windows: 1–2 h ahead, 1–4 h ahead and 1–6 h ahead. In each window, *Far* is the top-k mean of the forecast prices, where $k = 4$. The 1–2 h window contains only two trading intervals, so its *Far* is the average of those two.

3. **Weighted average cost.** At each 5-minute interval $t$:

$$
C_0 = p_0
$$

$$
C_t =
\begin{cases}
\dfrac{SOC_{t-1} \cdot C_{t-1} + E_t\ \cdot p_t}{SOC_{t-1} + E_t} & \text{if } P_t < 0 \text{ (charging)} \\[2ex]
C_{t-1} & \text{otherwise}
\end{cases}
$$

$$
E_t = -P_t \cdot \frac{5}{60}
$$

where;

$C_t$ is the average cost of the stored energy (\$/MWh)

$p_t$ is the actual price (\$/MWh)

$SOC_t$ is the state of charge (MWh)

$P_t$ is the battery's power (MW, positive when discharging)

$E_t$ is the energy bought in interval $t$ (MWh)

$t = 0$ is the first interval in which the battery is bidding and its SOC is below 10% and,

$p_0$ is price at that interval

### Comparison of ideal response model vs actual behaviour

We compare the ideal response with actual behaviour and put each interval into one of four groups:
1. **Agree:** the ideal response and actual behaviour match (both charge, both discharge or both hold)
2. **Held instead of acting:** the ideal response is to charge or discharge, but the battery held (power idle, within $\pm$ 5MW)
3. **Acted instead of holding:** the ideal response is to hold, but the battery charged or discharged
4. **Opposite direction:** the battery did the opposite of the ideal response (e.g. the ideal response is to charge, but the battery discharged)

Across the fleet, battery behaviour agrees with the rules about 40–49% of the time. The longer the *Far* window, the more often batteries agree with the rules.

| Rule | Far window | Agree | Held instead of acting | Acted instead of holding | Opposite direction |
|---|---|---:|---:|---:|---:|
| Rule 1 | 1–2h ahead | 41.6% | 19.5% | 25.6% | 13.3% |
| Rule 1 | 1–4h ahead | 46.7% | 19.8% | 21.9% | 11.6% |
| Rule 1 | 1–6h ahead | 48.8% | 20.0% | 20.1% | 11.2% |
| Rule 1+2 | 1–2h ahead | 40.4% | 11.9% | 40.4% | 7.3% |
| Rule 1+2 | 1–4h ahead | 44.9% | 13.8% | 33.4% | 7.8% |
| Rule 1+2 | 1–6h ahead | 47.0% | 15.6% | 28.5% | 8.9% |
