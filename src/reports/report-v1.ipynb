### Background and motivation

**1. AEMO's Approach to System Adequacy**
* AEMO already employs both long-term and short-term adequacy measures.
* Long-Term and Medium-Term PASA (Projected Assessment of System Adequacy) produce adequacy measurements over a multi-year horizon, such as EUE (Expected Unserved Energy), which informs long-term decisions on whether additional generation is needed.
* Short-Term PASA details system adequacy, generation, and Essential System Services (ESS) requirements at a half-hourly resolution over a one-week horizon.


**2. Policies on ESR (Energy Storage Resources)**
* Batteries are subject to ESROI (the timeframe during which they are obligated to be available) and ESROD (the duration for which they must remain available).

---

### Research Gaps

* Battery availability is not independent across intervals and cannot be "declared" like other generators; instead, it is strongly linked to the battery's SOC evolution over time.
* Although ST-PASA is designed to measure system adequacy in the short term, it does not measure system inadequacy within the hours-ahead range.
* By the end of the ESROI (4-6 hours), batteries would be depleted.
