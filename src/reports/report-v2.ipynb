# WA Big Batteries Short-Term Adequacy Flags

**Data.** 5-minute resolution for six electric storage resources (four Collie, two
Kwinana), September 2023 – August 2026.

**Scope.** Nine case days: eight selected from CC2T Review Consultation, plus one day flagged by
AEMO. State of charge, power, operational demand and energy price are plotted for
each.

**Peak ESROD.** 17:30–21:00 AWST, shaded on every day plot below.

---
## 1. Background

- AEMO measures system adequacy at two horizons. Long-Term and Medium-Term PASA
  produce multi-year measures such as Expected Unserved Energy (EUE). Short-Term PASA
  gives half-hourly detail over a one-week horizon.
- Electric storage resources carry an obligation interval (ESROI, the window in
  which they must be available) and an obligation duration (ESROD, how long they
  must sustain it).
- In 2024-2026, the Peak ESROD is at 17:30–21:00 across business days.
- ESROI and ESROD are certification concepts. They do not indicate, on the day,
  whether the fleet will still be able to produce at 20:00.

Sources: AEMO, [Amendment to the Mid Peak ESROI 2025-26](https://www.aemo.com.au/-/media/files/electricity/wem/planning_and_forecasting/esoo/2025/amendments-to-mid-peak-esroi.pdf);
[2025 WEM ESOO](https://www.aemo.com.au/-/media/files/electricity/wem/planning_and_forecasting/esoo/2025/2025-wem-electricity-statement-of-opportunities.pdf);
[2025 Reserve Capacity Information Pack](https://www.aemo.com.au/-/media/files/electricity/wem/planning_and_forecasting/esoo/2025/2025-reserve-capacity-information-pack.pdf).

## 2. Research gap

Battery availability is not independent across intervals. It depends on the
resource's state of charge, which is a function of what it did earlier in the day,
so it cannot be declared interval by interval in the way a power plant unit's can.


- ST-PASA looks a week ahead in half-hour steps. A battery can go from fine to
  empty over one afternoon, so a weekly forecast doesn't see it.
- No measure currently reports how often depletion arrives before the demand peak
  has passed.
- What matters is the timeframe for when they run out, and it hasn't been taken
  into account. Empty at 21:00 is acceptable. Empty at 19:30, while demand is
  still peaking, is not. Nothing tracks which of those is happening, or how often.
- Another thing that matters is: if we know a stress event is coming, how early in
  advance do we need to know? For example, if a stress event happens at 18:00 and
  we only find out an hour before, we probably can't charge all the batteries in
  time. So how early do we need to know in order to prepare them?

## 3. Data and assumptions

- Alinta is excluded from analysis as it commenced operation in July 2026. The four Collie and two
  Kwinana units should be sufficient for analysing historical data.
- Participants' declared availability may be confidential and is not presented here.
- Battery charge/discharge power is expressed as a percentage of rated capacity. Positive is discharge,
  negative is charge.

## 6. Interpretation

Availability for an energy-limited resource is a property of a trajectory rather
than of an interval. A resource that is available at 17:30 may be physically
incapable of producing at 19:30 as a direct consequence of how it was operated in
the preceding hours.

> Nothing in the current short-term adequacy framework observes the state variable
that determines this: state of charge entering the window, and the rate at which
it is being spent.

---
## 7. Data quality

State of charge readings of exactly zero could imply three distinct
conditions: a genuinely empty resource, missing telemetry, and outage.

On 6 March 2025, KWINANA_ESR2 reads 0% for thirteen and a half hours and
then rises to 70% within a few intervals, which is not physically possible.

![](../../data/plots/2025-03-06_day.png)

*6 March 2025. KWINANA_ESR2 (brown) is at zero for thirteen and a half hours before rising vertically. This may come from a telemetry error*

![](../../data/plots/timeframe_soc_distribution_grid.png)

*State of charge distributions. Fleet means fall between 33% and 51%. The spike at exactly zero appears in all six panels and is potentially a combination of outages and artefact described above.*
