## Executive Summary

The WA battery fleet now meets about 15% of evening-peak demand, three times the share of a year earlier. On days where AEMO has identified a tight supply/demand balance, we observe batteries generally prepared and acted to support system security through maintaining charge before the system peak and discharging to manage it. The risks we see are emerging, not yet binding, and include: weaker charge in winter, batteries cancelling each other out just before the peak, and evening price forecasts that err high.

- Reliance has tripled. Batteries met 5.1% of demand in the evening obligation window (ESROI, 17:30–21:30) in Dec 2024–Sep 2025, and 14.6% in Dec 2025–Sep 2026. Evening demand fell 3% over the same period, so the rise comes from more batteries, used harder.
- Charge at the peak is almost always adequate. Median fleet state of charge (SOC) at 17:30 is 71%. It is below 50% on 11% of days and below 40% on 4%. On the 7 CC2T and 14 AEMO-advisory stress days with reliable data, SOC sat within the normal range.
- But usually below what credits assume. Delivering credited capacity for 4 hours needs about 78% fleet SOC. The median day falls short of that, and winter 2026 dipped to a median of about 45% (May–June).
- Behaviour largely follows the forecasts. In the evening window, batteries' cheapest discharge offer sits close to the best price forecast for the rest of the window (median +\$1/MWh). Under the ERA bidding guideline, this is expected behaviour.
- Owners behave differently. Synergy (regulated) offers discharge at about \$115/MWh (noting this doesn’t account for the bid stack), runs a morning cycle and provides most FCESS. Neoen offers at about \$239/MWh and cycles once a day. Synergy's batteries set most WEM prices.
- Batteries now set WEM prices more than 90% of the time. Forecasts, offers and outcomes are therefore circular: battery behaviour shapes the prices batteries respond to.
- Evening forecasts are less reliable than midday ones. Midday forecast error is stable with lead time.Evening forecasts over-estimate price increasingly the further ahead they look.

**Implications for next steps for this project:** The data show no evidence of a systemic energy shortfall to date. D3 will develop indicators of storage adequacy risk, including risk that may not be evident from operational forecasts and market advisories. It will also identify the conditions under which that risk is likely to increase: growth of the battery fleet beyond the available midday charging energy, coal retirement tightening winter evening peaks, and price forecast error interacting with narrowing intraday price spreads.

### How synchronised are the batteries?

The share of the batteries' activity that cancels out (some charging while others discharging) instead of being used by the grid:

$$
\text{Cancelling out (\%)} = \left(1 - \frac{\text{Energy reaching the grid}}{\text{All battery activity}}\right) \times 100
$$

- **Energy reaching the grid (MWh):** in each 5-minute interval, the absolute value of fleet's total discharging minus its total charging, converted to energy (MW × 5 minutes = MWh) and added up over the month (or hour of day).
- **All battery activity (MWh):** in each 5-minute interval, the fleet's total discharging plus its total charging, converted to MWh and added up the same way.

For example, if for one hour battery A discharges 100 MW while battery B charges 30 MW, total activity is 130 MW and 70 MW reaches the grid (does not cancel out), so 1 − 70 ÷ 130 = 46% cancels out.

# D3 - Storage Reserve Adequacy Risk Indicators

## Understanding batteries' ideal response to the price forecast

To model how batteries will behave in the future, we started by examining how batteries currently respond to the price forecast. The Offer Construction Guideline allows ESRs to follow an optimal pricing strategy that accounts for their opportunity cost and their obligation to preserve energy for the ESROI window.

### Ideal response modelling

As the Offer Construction Guideline is open to interpretation, we propose the following model of a battery's ideal response to the price forecast.

- The AEMO pre-dispatch price forecast is divided into two categories: the *Near* forecast and the *Far* forecast. The *Near* forecast covers the next hour, whereas the *Far* forecast is a set of windows: 1–2 h ahead, 1–4 h ahead and 1–6 h ahead.
- Comparing the *Near* and *Far* forecasts forms the basis for determining a battery's ideal response.
- The weighted average cost of the energy stored in the battery is another factor in determining the ideal response. We therefore propose two rules to model a battery's ideal response:

1. **Rule 1: the battery acts only on the comparison of the *Near* and *Far* forecasts**
    - Discharge: *Near* > *Far*
    - Charge: *Near* < 0.85 $\cdot$ *Far*, where 0.85 is the round-trip efficiency
    - Hold: otherwise
2. **Rule 2: the battery acts on the price forecast and the weighted average cost**
    - Discharge: *Near* > *Far* **and** *Near* > *Avg cost*
    - Charge: *Near* < 0.85 $\cdot$ *Far*, where 0.85 is the round-trip efficiency
    - Hold: otherwise

*Near*, *Far* and *Avg cost* are defined as follows.

1. **Near forecast.** For each trading interval, *Near* is the average forecast price over the next hour (the next two trading intervals). It uses the forecast published one trading interval earlier.

2. **Far forecast.** For each trading interval, *Far* is calculated for three forecast windows: 1–2 h ahead, 1–4 h ahead and 1–6 h ahead. In each window, *Far* is the top-k mean of the forecast prices, where $k = 4$. The 1–2 h window contains only two trading intervals, so its *Far* is the average of those two.

3. **Weighted average cost.** At each 5-minute interval $t$:

$$
C_0 = p_0
$$

$$
C_t =
\begin{cases}
\dfrac{SOC_{t-1} \cdot C_{t-1} + E_t\ \cdot p_t}{SOC_{t-1} + E_t} & \text{if } P_t < 0 \text{ (charging)} \\[2ex]
C_{t-1} & \text{otherwise}
\end{cases}
$$

$$
E_t = -P_t \cdot \frac{5}{60}
$$

where;

$C_t$ is the average cost of the stored energy (\$/MWh)

$p_t$ is the actual price (\$/MWh)

$SOC_t$ is the state of charge (MWh)

$P_t$ is the battery's power (MW, positive when discharging)

$E_t$ is the energy bought in interval $t$ (MWh)

$t = 0$ is the first interval in which the battery is bidding and its SOC is below 10% and,

$p_0$ is price at that interval

### Comparison of ideal response model vs actual behaviour

The ideal response vs actual outcome is categorised into 4 groups:
1. **Agree:** both ideal response and actual are either charge/discharge/hold
2. **Held instead of acting:** ideal response says either charge/discharge, whereas actual is hold (power is idle, within $\pm$ 5MW)
3. **Acted instead of holding:** ideal response says hold, whereas actual is either charge/discharge
4. **Opposite direction:** actual is the opposite direction of what the ideal response says (eg. ideal response says charge but actual is discharge)

Across the whole fleet, batteries behaviour is aligned with the rules at around 40-48% of the time. The farther ahead the forecast, the more batteries agree with the rule.

| Rule | Far window | Agree | Held instead of acting | Acted instead of holding | Opposite direction |
|---|---|---:|---:|---:|---:|
| Rule 1 | 1–2h ahead | 41.6% | 19.5% | 25.6% | 13.3% |
| Rule 1 | 1–4h ahead | 46.7% | 19.8% | 21.9% | 11.6% |
| Rule 1 | 1–6h ahead | 48.8% | 20.0% | 20.1% | 11.2% |
| Rule 1+2 | 1–2h ahead | 40.4% | 11.9% | 40.4% | 7.3% |
| Rule 1+2 | 1–4h ahead | 44.9% | 13.8% | 33.4% | 7.8% |
| Rule 1+2 | 1–6h ahead | 47.0% | 15.6% | 28.5% | 8.9% |

### The rules each battery follows

Adding **Rule 2** improves NEOEN batteries' compliance with the ideal response, whereas it reduces Synergy batteries' compliance. This suggests that NEOEN facilities may take into account their weighted average cost as a decision factor, whereas Synergy facilities may take into account only the price forecast comparison.
