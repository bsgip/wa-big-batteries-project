## WA Energy System Adequacy Analysis - Grid-Scale Batteries

## 1. Full History Fleet Analysis

Over the past 3 years (2023-2026), the emergence of grid-scale batteries has changed the SWIS. As of now, the total fleet capacity reaches up to 5,967 MWh.

Source: https://explore.openelectricity.org.au/facilities/wem/?tech=battery_discharging&status=operating

![](../../data/plots/fleet_capacity_evolution.png)

### Fleet Data Profile

Overall, the battery fleet has a mean SOC of 41.2%, with a standard deviation of 23%, and a median of 40.5%. At ESROI start time (17:30), both averages are higher: a mean of 59%, a standard deviation of 22%, and a median of 65%. This indicates that the battery fleet holds a higher SOC at peak times than its overall average.

The SOC distribution, however, is not uniform around its median. Three distinct peaks appear instead, which may indicate clustering based on different battery facilities. This will be investigated further in the battery comparison section.

![](../../data/plots/fleet_soc_distribution.png)
![](../../data/plots/fleet_soc_distribution_esroi_start.png)

**At ESROI start, 8.9% of the SOC values are 20% or below**, and **18.9% are 40% or below**. Those 18.9% break down per battery as follows.

### Seasonal Fleet Profile

Interestingly, the median SOC is slightly higher in summer than in winter. Winter also shows two distribution peaks, one of them a cluster of SOC below 20%. This needs to be investigated further.

![](../../data/plots/fleet_soc_distribution_by_season.png)

## 2. Evolution of the system & stress events

Consistent with the seasonal comparison, the trend throughout the year appears to show that the fleet SOC at ESROI start is lower in the winter months, particularly in 2024 and 2026. Summer 2024 saw quite a significant drop in fleet SOC, which might be due to SSE events, whereas summer 2025 did not, despite having a few stress events.

![](../../data/plots/fleet_soc_evolution_trend_events.png)

![](../../data/plots/fleet_soc_evolution_trend_seasonal_events.png)

![](../../data/plots/sse_category_pie_chart.png)

Most stress events occur within one day, however, there were 3 events that occurred within more than 10 days, with the longest one being 157 days in May to October 2024.

![](../../data/plots/LRC_dates.png)

### Past Year Analysis

Over the past year, the fleet SOC distribution is a uniform normal distribution with a mean of 46.6%, higher than the full history mean of 41.2%. At ESROI start, the fleet SOC seems to concentrate at around 65%, with a mean of 67.7% and a standard deviation of 14%.

![](../../data/plots/fleet_soc_distribution_past_year.png)

## 4. Battery Behaviour Analysis

From a general point of view, the heatmap shows that each battery behaves as expected: charging during the day and discharging during peak time. This indicates that the batteries have played a role in reducing stress during peak time.

![](../../data/plots/summary-plot/timeframe_power_heatmap_grid.png)

The box-and-whisker plot for each battery indicates that NEOEN and Synergy batteries follow different charging/discharging strategies, with Synergy batteries typically discharging in the morning, around 6-7am.
