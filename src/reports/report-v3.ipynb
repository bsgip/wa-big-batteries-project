## WA Energy System Adequacy Analysis - Grid-Scale Batteries

## 1. Full History Fleet Analysis

Over the past 3 years (2023-2026), the emergence of grid-scale batteries has changed the SWIS. As of now, the total fleet capacity reaches up to 5,967 MWh.

Source: https://explore.openelectricity.org.au/facilities/wem/?tech=battery_discharging&status=operating

![](../../data/plots/fleet_description/fleet_capacity_evolution.png)

### Fleet Data Profile

Data of the batteries fleet are available from 2023. However, due to outages and missing telemetry during the early stage of commissioning, we start our analysis from **December 2024**.

![](../../data/plots/data_description/data_period_overview.png)

Overall, the battery fleet has a mean SOC of 47.2%, with a standard deviation of 19.3%, and a median of 46.1%. At ESROI start time (17:30), both averages are higher: a mean of 66.1%, a standard deviation of 15.3%, and a median of 69.6%. This indicates that the battery fleet holds a higher SOC at peak times than its overall average.

The SOC distribution at ESROI start, however, is not uniform around its median. Two distinct peaks appear instead, which may indicate clustering potentially based on seasons, different battery facilities, and other factors. This will be investigated further.

![](../../data/plots/fleet_description/fleet_soc_distribution.png)
![](../../data/plots/fleet_description/fleet_soc_distribution_esroi_start.png)

**At ESROI start (17:30), the fleet's SOC was never 20% or below**, and it was **40% or below on 9.1% of days**. The chart below shows which batteries had low SOC (40% or below) at ESROI start.

## 4. Battery Behaviour Analysis

From a general point of view, the heatmap shows that each battery behaves as expected: charging during the day and discharging during peak time. This indicates that the batteries have played a role in reducing stress during peak time.

![](../../data/plots/batteries_comparison/timeframe_power_heatmap_grid.png)

The box-and-whisker plot for each battery indicates that NEOEN and Synergy batteries follow different charging/discharging strategies, with Synergy batteries typically discharging in the morning, around 6-7am.
