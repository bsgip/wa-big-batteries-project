### What are the batteries reserve capacity obligations?

AEMO publishes the Reserve Capacity Obligation Quantity (RCOQ) for the batteries for each dispatch interval. If we calculate how much SOC a battery needs according to this MW value, we can get its "credit equivalent" in MWh.

| Battery | Owner | Battery capacity (MWh) | RCOQ (MW) | Credit equivalent (MWh) | Credit equivalent as % of capacity | Obligation start |
|---| --- | ---|---|---|---|---|
| COLLIE_BESS2 | NEOEN | 1,363 | 300 | 1,200 | 88% | Jun 2026 |
| COLLIE_ESR1 | NEOEN | 877 | 200 | 800 | 91% | Jun 2025 |
| COLLIE_ESR4 | Synergy | 1,200 | 176.01 | 704 | 59% | Jun 2026 |
| COLLIE_ESR5 | Synergy | 1,200 | 175.35 | 701 | 58% | Jun 2026 |
| KWINANA_ESR1 | Synergy | 227 | 44.25 | 177 | 78% | Oct 2023 |
| KWINANA_ESR2 | Synergy | 900 | 225 | 900 | 100% | Jun 2026 |

*Credit equivalent = RCOQ (MW) × 4h ESROI window. RCOQ is derived from each battery's most common non-zero value in the 2025-26 capacity year (from 1 Oct 2025).*

Based on the derived credit equivalent as \% of capacity, the value may not be a fair measurement for how much SOC each battery needs at the start of ESROI, as some of them are within a high range of 80-100% SOC.

### Do batteries meet their capacity obligation at each dispatch interval?

To satisfy their obligation, battery has to make their RCOQ capacity available through the bid offers quantity. There is no specific guideline for **how much SOC a battery needs to have** in order to meet their obligation.

![](../../data/plots/batteries_comparison/offered_vs_rcoq_esroi_per_battery.png)

From the above graph, it can be observed that NEOEN batteries (COLLIE_BESS2 and COLLIE_ESR1) typically have offer quantities that are only slightly less than their RCOQ. Synergy's COLLIE_ESR4 and COLLIE_ESR5 typically offer above their RCOQ, and this might be because their RCOQ is only around 58% of their full capacity. Synergy's KWINANA_ESR1, which is the oldest battery in WA, had many days where it did not bid all of their obligation quantity in their offer.

## Defining stress days
Analysing system adequacy requires us to understand what counts as a "stress day". This can be explored from multiple perspectives. So far, we have at least 3 definition for when stress days occurred in the past:

- CC2T review: listed 8 stress event dates
- AEMO's market advisory LRC: 61 advisories, on 48 dates
- Intervals with spare capacity below 750MW: 35 dates

*Spare capacity = IN-SERVICE capacity - energy injection dispatch*

### Is the frequency of stress days changing over time based on each definition?

Both AEMO market advisory and CC2T review data show that summer months have the highest number of stress days.

![](../../data/plots/stress_events/stress_days_by_season.png)

### What time does a stress event typically start?

There is a lot of overlap in the strees event start hour of around 3-4 pm, which is consistent with the time a couple hours prior to peak time and ESROI start. On the other hand, the spare-capacity stress days mostly start at 5-6 pm, indicating that spare capacity can be needed more as peak demand increases during ESROI.

![](../../data/plots/stress_events/stress_days_by_season_hour.png)

Stress days according to a spare capacity below 750MW stopped occurring from summer 2024 onwards, which can be associated with the fact that IN-SERVICE capacity and spare capacity are both increasing over time.

![](../../data/plots/stress_events/in_service_capacity_trend.png)

![](../../data/plots/stress_events/spare_capacity_trend.png)

This is an interesting finding, because it turns out that the demand itself is only growing slowly. The average demand is increasing about 2-4% a year.

### How synchronised are the batteries?

The share of the batteries' activity that cancels out (some charging while others discharging) instead of being used by the grid:

$$
\text{Cancelling out (\%)} = \left(1 - \frac{\text{Energy reaching the grid}}{\text{All battery activity}}\right) \times 100
$$

- **Energy reaching the grid (MWh):** in each 5-minute interval, the absolute value of fleet's total discharging minus its total charging, converted to energy (MW × 5 minutes = MWh) and added up over the month (or hour of day).
- **All battery activity (MWh):** in each 5-minute interval, the fleet's total discharging plus its total charging, converted to MWh and added up the same way.

For example, if for one hour battery A discharges 100 MW while battery B charges 30 MW, total activity is 130 MW and 70 MW reaches the grid (does not cancel out), so 1 − 70 ÷ 130 = 46% cancels out.
