```python
df = pd.read_csv("wemde_fields.csv")
df.head()
```

```python
search_terms = [
    "bess",
    "ess",
    "esr",
    "battery",
    "capacity",
    "storage",
    "charge",
    "soc",
    "mwh",
    "energy",
    "bidirectional"
]
```

```python
def matches(path):
    p = path.lower()
    return any(t in p for t in search_terms)

result = df[df["path"].apply(matches)]
result.head()
```
